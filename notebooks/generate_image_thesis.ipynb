```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path


def plot_persta_throughput_boxplot(
    nodedensity_csv: str | Path,
    save_dir: str | Path,
    filename: str = "Fig4_1_perSTA_throughput.png",
    showImage: bool = False,
):
    """
    Generate Figure 4.1:
    Per-STA throughput distribution grouped by node density.
    """
    save_dir = Path(save_dir)
    save_dir.mkdir(parents=True, exist_ok=True)

    df = pd.read_csv(nodedensity_csv)

    plt.figure(figsize=(7, 5))
    df.boxplot(
        column="AvgThroughput(Mbps)",
        by="NodeDensity",
        grid=False,
    )

    plt.title("Per-STA Throughput Distribution by Node Density")
    plt.suptitle("")  # remove pandas auto-title
    plt.xlabel("Node Density")
    plt.ylabel("Throughput (Mbps)")

    plt.tight_layout()
    plt.savefig(save_dir / filename, dpi=300)
    if showImage:
        plt.show()
    plt.close()


def plot_persta_rssi_distribution(
    rssi_csv: str | Path,
    save_dir: str | Path,
    filename: str = "Fig4_2_perSTA_RSSI.png",
    bins: int = 40,
    showImage: bool = False,
):
    """
    Generate Figure 4.2:
    Histogram of per-STA RSSI measurements.
    """
    save_dir = Path(save_dir)
    save_dir.mkdir(parents=True, exist_ok=True)

    df = pd.read_csv(rssi_csv)

    plt.figure(figsize=(7, 5))
    plt.hist(df["signal_dBm"], bins=bins)

    plt.xlabel("RSSI (dBm)")
    plt.ylabel("Frequency")
    plt.title("Distribution of Per-STA RSSI Measurements")

    plt.tight_layout()
    plt.savefig(save_dir / filename, dpi=300)
    if showImage:
        plt.show()
    plt.close()


# -------------------------------------------------
# Example usage (Chapter 4 figures)
# -------------------------------------------------
# if __name__ == "__main__":
#     BASE_DIR = Path("outputs/csv/wifi-random")

#     plot_persta_throughput_boxplot(
#         nodedensity_csv=BASE_DIR / "wifi-random-11-Dec-2025_11-55-seed10-nodedensity.csv",
#         save_dir="figures",
#     )

#     plot_persta_rssi_distribution(
#         rssi_csv=BASE_DIR / "wifi-random-11-Dec-2025_11-55-seed10-rssi.csv",
#         save_dir="figures",
#     )

#     print("Chapter 4 figures generated successfully.")
```
